```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from pathlib import Path

from src.task import DeepLSTM, preparar_dados

plt.rcParams.update({
    'figure.dpi': 120, 'axes.spines.top': False, 'axes.spines.right': False,
    'axes.grid': True, 'grid.alpha': 0.3, 'font.size': 11
})

DEVICE       = 'cuda' if torch.cuda.is_available() else 'cpu'
CORES        = {'centralizado':'#4CAF50','local_0':'#9C27B0','local_1':'#673AB7','fedavg':'#2196F3','fedprox':'#FF5722'}
DADOS_DIR = Path(r"C:\TCC\prova_conceito\dados")
RESULTS_DIR  = Path('results');  RESULTS_DIR.mkdir(exist_ok=True)
N_CLIENTES   = 2
EPOCAS_TOTAL = 30
BATCH_SIZE   = 64
LR           = 0.0005
N_HORAS_VIS  = 500
JANELA       = 24

N_FEATURES = 3  # Apenas a temperatura como entrada da LSTM

# ── Ler os Excels e montar dados_clientes ──────────────────────────────────
dados_clientes = []

for i in range(N_CLIENTES):
    caminho = DADOS_DIR / f'estacao_{i}.csv'
    df = pd.read_csv(caminho)
    X_train, y_train, X_test, y_test, escala = preparar_dados(df)

    dados_clientes.append({
        'X_train': X_train, 'y_train': y_train,
        'X_test':  X_test,  'y_test':  y_test,
        't_min':   escala[0],
        't_max':   escala[1],
    })
    print(f'Estação {i} → treino: {X_train.shape} | teste: {X_test.shape} | '
          f'T∈[{escala[0]:.1f}, {escala[1]:.1f}]°C')

print('\n✅ dados_clientes montado!')
```

---
## Célula 4 — Funções auxiliares
(idêntica ao notebook original)

```python
def treinar_modelo(X_train, y_train, epochs=30, lr=0.001, batch_size=64, desc=''):
    modelo = DeepLSTM().to(DEVICE)
    opt    = torch.optim.Adam(modelo.parameters(), lr=lr)
    crit   = nn.MSELoss()
    loader = torch.utils.data.DataLoader(
        torch.utils.data.TensorDataset(X_train, y_train),
        batch_size=batch_size, shuffle=True)
    historico = []
    for epoch in range(epochs):
        modelo.train()
        total = 0.0
        for Xb, yb in loader:
            Xb, yb = Xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            loss = crit(modelo(Xb), yb)
            loss.backward()
            opt.step()
            total += loss.item()
        historico.append(total / len(loader))
        if (epoch + 1) % 10 == 0:
            print(f'  {desc} | Época {epoch+1:3d}/{epochs} | Loss: {historico[-1]:.6f}')
    return modelo, historico

def avaliar(modelo, dc):
    if modelo is None: return None, None
    modelo.eval()
    t_min, t_max = dc['t_min'], dc['t_max']
    with torch.no_grad():
        p = modelo(dc['X_test'].to(DEVICE)).cpu().detach().numpy().squeeze()
    r = dc['y_test'].detach().numpy()
    rmse_n = float(np.sqrt(np.mean((p - r) ** 2)))
    p_c    = p * (t_max - t_min) + t_min
    r_c    = r * (t_max - t_min) + t_min
    return rmse_n, float(np.sqrt(np.mean((p_c - r_c) ** 2)))


def predicoes_desnorm(modelo, dc, n=168):
    if modelo is None: return None, None
    modelo.eval()
    t_min, t_max = dc['t_min'], dc['t_max']
    with torch.no_grad():
        p = modelo(dc['X_test'][-n:].to(DEVICE)).cpu().detach().numpy().squeeze()
    r = dc['y_test'][-n:].detach().numpy()
    return p*(t_max-t_min)+t_min, r*(t_max-t_min)+t_min


def carregar_fed(strategy):
    arq = RESULTS_DIR / f'melhores_pesos_{strategy}.pt'
    if not arq.exists(): print(f'Não encontrado: {arq}'); return None
    ckpt = torch.load(arq, map_location='cpu', weights_only=False)
    m = DeepLSTM(); m.load_state_dict(ckpt['model_state_dict']); m.eval()
    print(f'{strategy.upper()} | Rodada {ckpt["rodada"]} | RMSE: {ckpt["rmse"]:.4f}')
    return m


def carregar_metricas(strategy):
    arq = RESULTS_DIR / f'metricas_{strategy}.json'
    return json.loads(arq.read_text(encoding='utf-8')) if arq.exists() else None


# Constante usada em erro_por_hora (N_HORAS definida na Célula 2)
N_HORAS = 8760

print('✅ Funções auxiliares prontas!')
```

---
## Célula 5 — Treinamento Centralizado

Concatena os dados das duas estações e treina **um único modelo** com tudo.

```python
X_central = torch.cat([d['X_train'] for d in dados_clientes])
y_central = torch.cat([d['y_train'] for d in dados_clientes])
print(f'Dataset centralizado: {len(X_central)} amostras\n')

modelo_central, loss_central = treinar_modelo(
    X_central, y_central,
    epochs=EPOCAS_TOTAL, lr=LR, batch_size=BATCH_SIZE,
    desc='Centralizado'
)

torch.save(modelo_central.state_dict(), RESULTS_DIR / 'pesos_centralizado.pt')
print('\n✅ Modelo centralizado salvo!')
```

---
## Célula 6 — Treinamento Local Isolado

Cada estação treina seu próprio modelo **sem ver os dados da outra**.

```python
modelos_locais = []
losses_locais  = []

for i, dc in enumerate(dados_clientes):
    print(f'\n--- Estação {i} ---')
    m, h = treinar_modelo(
        dc['X_train'], dc['y_train'],
        epochs=EPOCAS_TOTAL, lr=LR, batch_size=BATCH_SIZE,
        desc=f'Estacao {i}'
    )
    modelos_locais.append(m)
    losses_locais.append(h)
    torch.save(m.state_dict(), RESULTS_DIR / f'pesos_local_{i}.pt')

print('\n✅ Modelos locais salvos!')
```

---
## Célula 7 — Carregar modelo Federado (FedAvg)

Se você ainda não rodou o Flower, `modelo_fedavg` ficará `None`  
e os gráficos seguintes simplesmente pulam a curva federada — sem quebrar.

```python
modelo_fedavg = carregar_fed('fedavg')
dados_fedavg  = carregar_metricas('fedavg')
```

---
## Célula 8 — Tabela comparativa de RMSE
(idêntica ao notebook original)

```python
linhas = []
for i, dc in enumerate(dados_clientes):
    for nome, modelo in [
        ('Centralizado',  modelo_central),
        ('Local Isolado', modelos_locais[i]),
        ('FedAvg',        modelo_fedavg),
    ]:
        rn, rc = avaliar(modelo, dc)
        linhas.append({'Paradigma': nome, 'Estacao': i, 'RMSE norm': rn, 'RMSE (C)': rc})

df_res = pd.DataFrame(linhas)
tabela = (
    df_res.groupby('Paradigma')[['RMSE norm','RMSE (C)']]
    .agg(['mean','std']).round(4)
    .sort_values(('RMSE (C)','mean'))
)
tabela.columns = ['RMSE norm média','RMSE norm std','RMSE °C média','RMSE °C std']

print('=== TABELA COMPARATIVA ===')
print(tabela.to_string())
tabela.to_csv(RESULTS_DIR / 'comparativo_paradigmas.csv')
```

---
## Célula 9 — Gráfico de barras — RMSE por paradigma

```python
paradigmas = ['Centralizado','Local Isolado','FedAvg']
cores_b    = [CORES['centralizado'], CORES['local_0'], CORES['fedavg']]

fig, axes = plt.subplots(1, N_CLIENTES, figsize=(7*N_CLIENTES, 5), sharey=True)
if N_CLIENTES == 1: axes = [axes]

for i, ax in enumerate(axes):
    sub  = df_res[df_res['Estacao'] == i]
    vals = [sub[sub['Paradigma']==p]['RMSE (C)'].values for p in paradigmas]
    vals = [v[0] if len(v) and v[0] is not None else 0 for v in vals]
    bars = ax.bar(paradigmas, vals, color=cores_b, alpha=0.85, edgecolor='white', linewidth=1.5)
    for bar, val in zip(bars, vals):
        if val:
            ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+0.02,
                    f'{val:.3f}°C', ha='center', va='bottom', fontsize=9, fontweight='bold')
    ax.set_title(f'Estação {i}'); ax.set_ylabel('RMSE (°C)'); ax.tick_params(axis='x', rotation=15)

plt.suptitle('RMSE por Paradigma', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'comparativo_barras.png', bbox_inches='tight')
plt.show()
```

---
## Célula 10 — Real vs Predito — todos os paradigmas

```python
for i, dc in enumerate(dados_clientes):
    mplot = [
        ('Centralizado',  modelo_central,    CORES['centralizado']),
        ('Local Isolado', modelos_locais[i], CORES[f'local_{i}']),
        ('FedAvg',        modelo_fedavg,     CORES['fedavg']),
    ]
    horas = np.arange(N_HORAS_VIS)
    fig, axes = plt.subplots(len(mplot), 1, figsize=(14, 4*len(mplot)))

    for ax, (nome, modelo, cor) in zip(axes, mplot):
        pc, rc = predicoes_desnorm(modelo, dc, N_HORAS_VIS)
        if pc is None:
            ax.set_title(f'{nome} — não disponível'); ax.axis('off'); continue
        rmse = float(np.sqrt(np.mean((pc-rc)**2)))
        ax.plot(horas, rc, color='#333333', lw=1.5, label='Real',    alpha=0.85)
        ax.plot(horas, pc, color=cor,       lw=1.5, label='Predito', alpha=0.9, linestyle='--')
        ax.fill_between(horas, rc, pc, alpha=0.1, color=cor)
        ax.set_title(f'{nome} | RMSE: {rmse:.4f}°C')
        ax.set_ylabel('Temp (°C)'); ax.legend(loc='upper right')

    axes[-1].set_xlabel('Hora')
    plt.suptitle(f'Estação {i} — Real vs Predito (últimas {N_HORAS_VIS}h)',
                 fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f'real_vs_predito_estacao_{i}.png', bbox_inches='tight')
    plt.show()
```

---
## Célula 11 — Convergência federada com referência centralizada

```python
fig, ax = plt.subplots(figsize=(10, 5))

for nome, dados in [('FedAvg', dados_fedavg)]:
    if dados is None:
        print(f'{nome}: métricas federadas não disponíveis — rode o Flower primeiro.')
        continue
    chave   = nome.lower()
    rodadas = [h['rodada']      for h in dados['historico']]
    rmse    = [h['rmse_global'] for h in dados['historico']]
    mr, mv  = dados['melhor_rodada'], dados['melhor_rmse']
    ax.plot(rodadas, rmse, marker='o', lw=2, color=CORES[chave], label=nome, markersize=5)
    ax.scatter([mr], [mv], s=150, zorder=5, color=CORES[chave],
               edgecolors='white', lw=2, label=f'Melhor {nome} (R{mr}: {mv:.4f})')

rmse_ref = df_res[df_res['Paradigma']=='Centralizado']['RMSE norm'].mean()
ax.axhline(rmse_ref, color=CORES['centralizado'], linestyle=':', lw=2,
           label=f'Centralizado ref: {rmse_ref:.4f}')
ax.set_xlabel('Rodada'); ax.set_ylabel('RMSE (norm)')
ax.set_title('Convergência Federada vs Referência Centralizada'); ax.legend()
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'convergencia_comparativa.png', bbox_inches='tight')
plt.show()
```

## Célula 13 — Resumo final para o TCC

```python
print('='*60)
print('RESUMO COMPARATIVO — TCC')
print('='*60)
print(f'Épocas equivalentes : {EPOCAS_TOTAL}')
print(f'Estações simuladas  : {N_CLIENTES}')
print()

medias = {}
for p in ['Centralizado','Local Isolado','FedAvg']:
    vals = df_res[df_res['Paradigma']==p]['RMSE (C)'].dropna()
    if len(vals) == 0:
        print(f'{p:<20}: não disponível'); continue
    media, std = vals.mean(), vals.std(); medias[p] = media
    por_est = ' | '.join([
        f'Est.{int(r["Estacao"])}: {r["RMSE (C)"]:.4f}°C'
        for _, r in df_res[df_res['Paradigma']==p].iterrows()
    ])
    print(f'{p:<20}: {media:.4f}°C ± {std:.4f} | {por_est}')

if medias:
    melhor = min(medias, key=medias.get)
    print(f'\nMelhor paradigma: {melhor} ({medias[melhor]:.4f}°C)')

arquivos = list(RESULTS_DIR.glob('*.png')) + list(RESULTS_DIR.glob('*.csv'))
print(f'\nArquivos gerados ({len(arquivos)}):')
[print(f'  {a.name}') for a in sorted(arquivos)]
print('='*60)
```

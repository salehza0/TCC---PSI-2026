# TCC — Visualização dos Resultados do Aprendizado Federado

Este notebook carrega os resultados salvos pela simulação Flower e gera:
1. **Curvas de convergência** — RMSE por rodada (FedAvg vs FedProx)
2. **Comparação entre clientes** — RMSE individual por estratégia
3. **Real vs Predito** — predições do modelo final nos dados de cada cliente
4. **Erro por hora do dia** — onde o modelo erra mais

> ⚠️ Execute a simulação com `flwr run . --stream` antes de rodar este notebook.
> Os arquivos `results/metricas_fedavg.json` e `results/melhores_pesos_fedavg.pt` devem existir.

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import torch
from pathlib import Path

# Adiciona o diretório raiz ao path para importar src
import sys
sys.path.insert(0, str(Path('.').resolve()))

from src.task import DeepLSTM, gerar_dados_sinteticos, preparar_dados, set_weights, evaluate

# Configuração visual
plt.rcParams.update({
    'figure.dpi': 120,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.3,
    'font.size': 11,
})

CORES = {'fedavg': '#2196F3', 'fedprox': '#FF5722'}
RESULTS_DIR = Path('results')
print('Bibliotecas carregadas com sucesso!')
```

## 1. Carregar métricas salvas pela simulação

```python
def carregar_metricas(strategy: str) -> dict | None:
    arquivo = RESULTS_DIR / f'metricas_{strategy}.json'
    if not arquivo.exists():
        print(f'⚠️  Arquivo não encontrado: {arquivo}')
        print(f'   Execute: flwr run . --stream --run-config "strategy={strategy}"')
        return None
    with open(arquivo, encoding='utf-8') as f:
        dados = json.load(f)
    print(f'✅ {strategy.upper()} | {len(dados["historico"])} rodadas | Melhor RMSE: {dados["melhor_rmse"]:.4f}°C (rodada {dados["melhor_rodada"]})')
    return dados

dados_fedavg  = carregar_metricas('fedavg')
dados_fedprox = carregar_metricas('fedprox')
```

## 2. Curva de convergência — RMSE por rodada

```python
fig, ax = plt.subplots(figsize=(10, 5))

for nome, dados in [('fedavg', dados_fedavg), ('fedprox', dados_fedprox)]:
    if dados is None:
        continue
    rodadas = [h['rodada'] for h in dados['historico']]
    rmse    = [h['rmse_global'] for h in dados['historico']]
    melhor_r = dados['melhor_rodada']
    melhor_v = dados['melhor_rmse']

    ax.plot(rodadas, rmse, marker='o', linewidth=2,
            color=CORES[nome], label=nome.upper(), markersize=5)
    ax.scatter([melhor_r], [melhor_v], s=150, zorder=5,
               color=CORES[nome], edgecolors='white', linewidth=2,
               label=f'Melhor {nome.upper()} (R{melhor_r}: {melhor_v:.4f}°C)')

ax.set_xlabel('Rodada de Comunicação')
ax.set_ylabel('RMSE (°C)')
ax.set_title('Convergência do Modelo Federado por Estratégia')
ax.legend()
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'convergencia.png', bbox_inches='tight')
plt.show()
print('Gráfico salvo em results/convergencia.png')
```

## 3. RMSE por cliente — comparação entre estratégias

```python
def extrair_rmse_clientes(dados: dict) -> dict:
    """Extrai a série de RMSE de cada cliente ao longo das rodadas."""
    clientes_series = {}
    for h in dados['historico']:
        for i, (cid, info) in enumerate(h['clientes'].items()):
            label = f'Estação {i}'
            if label not in clientes_series:
                clientes_series[label] = {'rodadas': [], 'rmse': []}
            clientes_series[label]['rodadas'].append(h['rodada'])
            clientes_series[label]['rmse'].append(info['rmse'])
    return clientes_series

n_graficos = sum(1 for d in [dados_fedavg, dados_fedprox] if d is not None)
fig, axes = plt.subplots(1, n_graficos, figsize=(7 * n_graficos, 5), sharey=True)
if n_graficos == 1:
    axes = [axes]

estilos_cliente = ['-', '--']
marcadores = ['o', 's']

for ax, (nome, dados) in zip(axes, [(n, d) for n, d in [('FedAvg', dados_fedavg), ('FedProx', dados_fedprox)] if d]):
    clientes = extrair_rmse_clientes(dados)
    for i, (label, series) in enumerate(clientes.items()):
        ax.plot(series['rodadas'], series['rmse'],
                linestyle=estilos_cliente[i % 2],
                marker=marcadores[i % 2],
                linewidth=2, markersize=5,
                label=label)
    ax.set_title(f'{nome} — RMSE por Estação')
    ax.set_xlabel('Rodada')
    ax.set_ylabel('RMSE (°C)')
    ax.legend()

plt.suptitle('Desempenho Individual dos Clientes', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'clientes_rmse.png', bbox_inches='tight')
plt.show()
print('Gráfico salvo em results/clientes_rmse.png')
```

## 4. Real vs Predito — modelo final em cada cliente

```python
def carregar_modelo(strategy: str) -> DeepLSTM | None:
    arquivo = RESULTS_DIR / f'melhores_pesos_{strategy}.pt'
    if not arquivo.exists():
        print(f'⚠️  Pesos não encontrados: {arquivo}')
        return None
    checkpoint = torch.load(arquivo, map_location='cpu', weights_only=False)
    modelo = DeepLSTM()
    modelo.load_state_dict(checkpoint['model_state_dict'])
    modelo.eval()
    print(f'✅ Pesos {strategy.upper()} carregados | Rodada {checkpoint["rodada"]} | RMSE: {checkpoint["rmse"]:.4f}°C')
    return modelo

def gerar_dados_cliente(partition_id: int):
    """Recria os dados exatamente como no cliente (mesma seed)."""
    np.random.seed(partition_id * 42)
    df = gerar_dados_sinteticos(n_amostras=8760, data_inicio=f'202{partition_id}-01-01')
    X_train, y_train, X_test, y_test, (t_min, t_max) = preparar_dados(df)
    return X_test, y_test, t_min, t_max

def predizer(modelo, X_test):
    with torch.no_grad():
        preds = modelo(X_test).squeeze().numpy()
    return preds

# Escolhe qual estratégia visualizar (altere para 'fedprox' se preferir)
STRATEGY_VIS = 'fedavg'
modelo = carregar_modelo(STRATEGY_VIS)

N_HORAS = 168  # última semana
N_CLIENTES = 2

if modelo:
    fig, axes = plt.subplots(N_CLIENTES, 1, figsize=(14, 5 * N_CLIENTES))
    if N_CLIENTES == 1:
        axes = [axes]

    for pid in range(N_CLIENTES):
        X_test, y_test, t_min, t_max = gerar_dados_cliente(pid)
        preds_norm = predizer(modelo, X_test[-N_HORAS:])
        real_norm  = y_test[-N_HORAS:].numpy()

        # Desnormaliza para graus Celsius
        preds_real = preds_norm * (t_max - t_min) + t_min
        real_real  = real_norm  * (t_max - t_min) + t_min

        horas = np.arange(N_HORAS)
        rmse_local = float(np.sqrt(np.mean((preds_real - real_real) ** 2)))

        ax = axes[pid]
        ax.plot(horas, real_real,  label='Real',    color='#333333', linewidth=1.5, alpha=0.85)
        ax.plot(horas, preds_real, label='Predito', color=CORES[STRATEGY_VIS],
                linewidth=1.5, linestyle='--', alpha=0.9)
        ax.fill_between(horas, real_real, preds_real, alpha=0.1, color=CORES[STRATEGY_VIS])
        ax.set_title(f'Estação {pid} — Últimas {N_HORAS}h | RMSE: {rmse_local:.4f}°C ({STRATEGY_VIS.upper()})')
        ax.set_xlabel('Hora')
        ax.set_ylabel('Temperatura (°C)')
        ax.legend()

    plt.suptitle('Real vs Predito — Modelo Federado Final', fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f'real_vs_predito_{STRATEGY_VIS}.png', bbox_inches='tight')
    plt.show()
    print(f'Gráfico salvo em results/real_vs_predito_{STRATEGY_VIS}.png')
```

## 5. Erro por hora do dia — onde o modelo erra mais

```python
if modelo:
    fig, axes = plt.subplots(1, N_CLIENTES, figsize=(7 * N_CLIENTES, 5), sharey=True)
    if N_CLIENTES == 1:
        axes = [axes]

    for pid in range(N_CLIENTES):
        X_test, y_test, t_min, t_max = gerar_dados_cliente(pid)
        preds_norm = predizer(modelo, X_test)
        real_norm  = y_test.numpy()

        preds_real = preds_norm * (t_max - t_min) + t_min
        real_real  = real_norm  * (t_max - t_min) + t_min
        erros_abs  = np.abs(preds_real - real_real)

        # Mapeia cada amostra de teste para sua hora do dia
        # (X_test começa após JANELA=24 amostras de treino)
        from src.task import JANELA
        n_total = 8760
        n_treino = int((n_total - JANELA) * 0.8)
        indices_teste = np.arange(n_treino, n_total - JANELA)
        horas_do_dia  = (indices_teste + JANELA) % 24

        erro_por_hora = pd.DataFrame({'hora': horas_do_dia[:len(erros_abs)], 'erro': erros_abs})
        media_por_hora = erro_por_hora.groupby('hora')['erro'].mean()

        ax = axes[pid]
        bars = ax.bar(media_por_hora.index, media_por_hora.values,
                      color=CORES[STRATEGY_VIS], alpha=0.8, edgecolor='white')
        ax.set_title(f'Estação {pid} — Erro Médio Absoluto por Hora do Dia')
        ax.set_xlabel('Hora do Dia')
        ax.set_ylabel('MAE (°C)')
        ax.set_xticks(range(0, 24, 2))

        # Destaca a hora com maior erro
        hora_pior = media_por_hora.idxmax()
        ax.bar(hora_pior, media_por_hora[hora_pior], color='#E53935', alpha=0.9,
               label=f'Pior hora: {hora_pior}h ({media_por_hora[hora_pior]:.3f}°C)')
        ax.legend()

    plt.suptitle(f'Análise de Erro por Hora — {STRATEGY_VIS.upper()}', fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / f'erro_por_hora_{STRATEGY_VIS}.png', bbox_inches='tight')
    plt.show()
    print(f'Gráfico salvo em results/erro_por_hora_{STRATEGY_VIS}.png')
```

## 6. Resumo final

```python
print('=' * 55)
print('RESUMO DOS EXPERIMENTOS')
print('=' * 55)

for nome, dados in [('FedAvg', dados_fedavg), ('FedProx', dados_fedprox)]:
    if dados is None:
        print(f'{nome}: dados não disponíveis')
        continue
    print(f'\n{nome}')
    print(f'  Rodadas         : {dados["total_rodadas"]}')
    print(f'  Melhor RMSE     : {dados["melhor_rmse"]:.4f}°C (rodada {dados["melhor_rodada"]})')
    rmse_final = dados['historico'][-1]['rmse_global']
    print(f'  RMSE final      : {rmse_final:.4f}°C')

if dados_fedavg and dados_fedprox:
    diff = dados_fedavg['melhor_rmse'] - dados_fedprox['melhor_rmse']
    vencedor = 'FedProx' if diff > 0 else 'FedAvg'
    print(f'\nMelhor estratégia : {vencedor} (diferenca: {abs(diff):.4f}°C)')

print('=' * 55)
print(f'Graficos salvos em: {RESULTS_DIR.resolve()}')
```
